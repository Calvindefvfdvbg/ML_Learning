```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
from sklearn.preprocessing import LabelEncoder
```

```python
# create a  data frame

data = pd.DataFrame({
    'Age': [25,35,45,20,50,30],
    'Income (K$/year)' : [40,60,80,30,100,50],
    'Credit Score': ['Low','Medium','High','Low','High','Medium'],
    'Loan Approved': ['No','Yes','Yes','No','Yes','No']
})
print(data)
```

```python
# encode categorial data
le = LabelEncoder()
data['Credit Score'] = le.fit_transform(data['Credit Score'])
x = data.drop('Loan Approved', axis = 1)
y = data['Loan Approved']
# Make Decision tree

model = DecisionTreeClassifier()
model.fit(x,y)


# Visualize tree
plt.figure(figsize=(10,6))
tree.plot_tree(model, feature_names = x.columns, class_names = ['No','Yes'], filled = True)
plt.show()

# Make prediction'Age=28, Income=55K, CreditScore=Medium (1)
new_data = [[28, 55, 1]]
prediction = model.predict(new_data)
print("Prediction:", "Yes" if prediction[0]==1 else "No")
```
