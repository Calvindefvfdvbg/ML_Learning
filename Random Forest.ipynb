```python
# python code to see bootstrap sampling in action

import numpy as np
# my candy lists

candies = ["Chocolate", "Gummy", "Lolipop", "Mint", "Caramel"]
print("original candies:", candies)

# pick three candies in randomly with replacement
sample = np.random.choice(candies, size=4, replace=True)
print("Random sample:", sample)
```

```python
# lets predicted if someone buys a phone based on age and income:

from sklearn.tree import DecisionTreeClassifier
import numpy as np

# fake data: [age, income], target(1 = buy, 0 = no buy)
x = np.array([[20,25000], [45,60000], [35,90000], [60,150000], [75,200000], [15,20000]])
y = np.array([0,1,0,1,1,0])

# build a trees with bagging
trees = []
for i in range(3):
     #bootstrap sample
     indices = np.random.choice(len(x), size=len(x), replace=True)
     x_sample = x[indices]
     y_sample = y[indices]
         
# Train a tree
tree = DecisionTreeClassifier()
tree.fit(x_sample, y_sample)
trees.append(tree)

# predict for a new person: age 15, income 35000
new_person = np.array([[15, 35000]])
votes = [tree.predict(new_person)[0] for tree in trees]
print("Tree votes:", votes)
final_guess = max(set(votes), key=votes.count)   # Majority vote
print("Bagging says:", "Buy" if final_guess==1 else "No buy")
```

```python
# let's  code a Random forest in python

from sklearn.ensemble import RandomForestClassifier

# Same data
x = np.array([[20,25000], [45,60000], [35,90000], [60,150000], [75,200000], [15,20000]])
y = np.array([0,1,0,1,1,0])

# Train a random forest 3 trees
rf = RandomForestClassifier(n_estimators = 3, random_state = 50)
rf.fit(x,y)

# predicted for our new person
new_person = np.array([[15, 35000]])
guess = rf.predict(new_person)
print("Random Forest says:", "Buy" if guess[0] == 1 else "No buy")
```

```python
from sklearn.metrics import accuracy_score
# Real vs. predicted
real = [0,1,0,1,1,0]
predicted = [0,0,0,1,1,0]
accuracy = accuracy_score(real,predicted)
print("Accuracy:", accuracy * 100, "%")
```
